```python
text_columns = ["Release Group", "Genres", "Original_Language", "Production_Countries"]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

print("Text columns cleaned successfully!")
```

```python
df["Rating"] = df["Rating"].astype(str).str.replace("/10", "", regex=False)
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")

print("Rating column converted successfully!")
```

```python
total_sales = df["$Worldwide"].sum()

print("Total Worldwide Sales:", total_sales)
```

```python
highest_sales = df["$Worldwide"].max()

print("Highest Worldwide Sales:", highest_sales)
```

```python
average_sales = df["$Worldwide"].mean()

print("Average Worldwide Sales:", average_sales)
```

```python
lowest_sales = df["$Worldwide"].min()

print("Lowest Worldwide Sales:", lowest_sales)
```

```python
total_movies = len(df)

print("Total Number of Movies:", total_movies)
```

```python
average_rating = df["Rating"].mean()

print("Average Rating:", average_rating)
```

```python
year_summary = df.groupby("Year")["$Worldwide"].agg(
    Number_of_Movies="count",
    Total_Worldwide_Sales="sum",
    Average_Worldwide_Sales="mean"
).sort_values("Total_Worldwide_Sales", ascending=False)

year_summary
```

```python
language_summary = df.groupby("Original_Language")["$Worldwide"].agg(
    Number_of_Movies="count",
    Total_Worldwide_Sales="sum",
    Average_Worldwide_Sales="mean"
).sort_values("Total_Worldwide_Sales", ascending=False)

language_summary
```

```python
country_summary = df.groupby("Production_Countries")["$Worldwide"].agg(
    Number_of_Movies="count",
    Total_Worldwide_Sales="sum",
    Average_Worldwide_Sales="mean"
).sort_values("Total_Worldwide_Sales", ascending=False)

country_summary.head(20)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
year_sales = df.groupby("Year")["$Worldwide"].sum()

plt.figure(figsize=(12, 6))
year_sales.plot(kind="bar")
plt.title("Total Worldwide Box Office Sales by Year")
plt.xlabel("Year")
plt.ylabel("Total Worldwide Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
year_domestic = df.groupby("Year")["$Domestic"].sum()
year_foreign = df.groupby("Year")["$Foreign"].sum()

plt.figure(figsize=(12, 6))
plt.plot(year_domestic.index, year_domestic.values, marker="o", label="Domestic Sales")
plt.plot(year_foreign.index, year_foreign.values, marker="o", label="Foreign Sales")

plt.title("Domestic vs Foreign Box Office Sales by Year")
plt.xlabel("Year")
plt.ylabel("Sales Value")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top_10_movies = df.sort_values("$Worldwide", ascending=False).head(10)

plt.figure(figsize=(12, 6))
plt.bar(top_10_movies["Release Group"], top_10_movies["$Worldwide"])

plt.title("Top 10 Movies by Worldwide Box Office Sales")
plt.xlabel("Movie")
plt.ylabel("Worldwide Sales")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
top_languages = language_summary.head(10)

plt.figure(figsize=(12, 6))
top_languages["Average_Worldwide_Sales"].plot(kind="bar")

plt.title("Average Worldwide Sales by Original Language")
plt.xlabel("Original Language")
plt.ylabel("Average Worldwide Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["Rating"].dropna(), bins=20, edgecolor="black")

plt.title("Distribution of Movie Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Movies")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top_countries = country_summary.head(10)

plt.figure(figsize=(12, 6))
top_countries["Total_Worldwide_Sales"].plot(kind="bar")

plt.title("Top 10 Production Countries by Worldwide Sales")
plt.xlabel("Production Country")
plt.ylabel("Total Worldwide Sales")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(df["Vote_Count"], df["$Worldwide"], alpha=0.5)

plt.title("Vote Count vs Worldwide Box Office Sales")
plt.xlabel("Vote Count")
plt.ylabel("Worldwide Sales")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
numeric_cols = ['Rank', '$Worldwide', '$Domestic', 'Domestic %', '$Foreign', 'Foreign %', 'Rating', 'Vote_Count']
numerical_df = df[numeric_cols]

# 4. Generate the correlation matrix
correlation_matrix = numerical_df.corr()

# Display the correlation matrix in the console
print("--- Correlation Matrix ---")
print(correlation_matrix)

# 5. Create the Heatmap Visual
plt.figure(figsize=(10, 8))
sns.heatmap(
    correlation_matrix, 
    annot=True,         
    cmap='coolwarm',     
    fmt=".2f",          
    linewidths=0.5,      
    vmin=-1, vmax=1      
)

plt.title('Movie Dataset Correlation Heatmap')
plt.tight_layout()

# Show the plot
plt.show()
```
